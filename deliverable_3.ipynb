## **Simulated Battles**

```python
print("Battle 1 - Character vs Character\n")
hero1 = Character("Hero1")
enemy1 = Character("Enemy1")

battle1 = Battle(hero1, enemy1)
winner1 = battle1.simulate_battle()
print(f"Winner: {winner1.name}\n")

print("Battle 2 - Character with Weapon vs Character\n")
sword = Weapon("Sword", Dice(20))

hero2 = Character("Hero2", weapon=sword)
enemy2 = Character("Enemy2")

battle2 = Battle(hero2, enemy2)
winner2 = battle2.simulate_battle()
print(f"Winner: {winner2.name}\n")

print("Battle 3 - Warrior vs Character\n")
hero3 = Warrior("Hero3")
enemy3 = Character("Enemy3")

battle3 = Battle(hero3, enemy3)
winner3 = battle3.simulate_battle()
print(f"Winner: {winner3.name}\n")

print("Battle 4 - Mage vs Warrior\n")
hero4 = Mage("Hero4")
enemy4 = Warrior("Enemy4")

battle4 = Battle(hero4, enemy4)
winner4 = battle4.simulate_battle()
print(f"Winner: {winner4.name}\n")

print("Battle 5 - Mage with Weapon vs Warrior with Weapon\n")
staff = Weapon("Staff", Dice(15))
axe = Weapon("Axe", Dice(21))

hero5 = Mage("Hero5", weapon=staff)
enemy5 = Warrior("Enemy5", weapon=axe)

battle5 = Battle(hero5, enemy5)
winner5 = battle5.simulate_battle()
print(f"Winner: {winner5.name}\n")
```
